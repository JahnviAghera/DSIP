```python
import numpy as np
import librosa

import matplotlib.pyplot as plt

# Load two mp3 audio files
audio_path1 = 'Vande Mataram (Maa Tujhe Salaam)-(Mr-Jat.in).mp3'
audio_path2 = 'Vande Mataram Karaoke -HQ.mp3'

y1, sr1 = librosa.load(audio_path1, sr=None, mono=True)
y2, sr2 = librosa.load(audio_path2, sr=None, mono=True)

# Ensure both signals have the same length
min_len = min(len(y1), len(y2))
y1 = y1[:min_len]
y2 = y2[:min_len]

# Cross-correlation
correlation = np.correlate(y1, y2, mode='full')

# Auto-correlation
autocorr1 = np.correlate(y1, y1, mode='full')
autocorr2 = np.correlate(y2, y2, mode='full')

# Plot results
plt.figure(figsize=(12, 8))

plt.subplot(3, 1, 1)
plt.title('Cross-correlation between audio1 and audio2')
plt.plot(correlation)
plt.xlabel('Lag')
plt.ylabel('Correlation')

plt.subplot(3, 1, 2)
plt.title('Auto-correlation of audio1')
plt.plot(autocorr1)
plt.xlabel('Lag')
plt.ylabel('Auto-correlation')

plt.subplot(3, 1, 3)
plt.title('Auto-correlation of audio2')
plt.plot(autocorr2)
plt.xlabel('Lag')
plt.ylabel('Auto-correlation')

plt.tight_layout()
plt.show()
```
