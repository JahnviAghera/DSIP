```python
from pydub import AudioSegment
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import butter, filtfilt

# Load MP3 file (make sure FFmpeg is installed and in your PATH)
audio = AudioSegment.from_mp3("BTS (방탄소년단) 'Euphoria _ Theme of LOVE YOURSELF 起 Wonder'.mp3")

# Convert stereo to mono and get raw samples
audio = audio.set_channels(1)
samples = np.array(audio.get_array_of_samples()).astype(np.float32)

# Define filter parameters for human voice range (approx. 80 Hz to 250 Hz for male, 140 Hz to 400 Hz for female, but a wider range is often used for general vocals)
# We'll use a general range, e.g., 100 Hz to 1000 Hz for demonstration
lowcut = 100  # Lower cutoff frequency in Hz
highcut = 1000 # Higher cutoff frequency in Hz
nyquist = 0.5 * audio.frame_rate
low = lowcut / nyquist
high = highcut / nyquist
order = 5

# Create the Butterworth filter
b, a = butter(order, [low, high], btype='band')

# Apply the filter
filtered_samples = filtfilt(b, a, samples)

# Normalize for plotting and export
filtered_samples_norm = filtered_samples / np.max(np.abs(filtered_samples))

# Scale to int16 for WAV export
filtered_samples_int16 = (filtered_samples_norm * 32767).astype(np.int16)

# Save filtered audio using PyDub
filtered_audio = AudioSegment(
    filtered_samples_int16.tobytes(),
    frame_rate=audio.frame_rate,
    sample_width=2,  # 2 bytes = 16 bits
    channels=1
)
filtered_audio.export("output_filtered_vocals.wav", format="wav")
print("Filtered audio saved as 'output_filtered_vocals.wav'.")

# Plot original and filtered audio
samples_norm = samples / np.max(np.abs(samples))  # normalize original for plotting

plt.figure(figsize=(12, 6))
plt.subplot(2, 1, 1)
plt.plot(samples_norm, color='blue')
plt.title("Original Audio Signal")

plt.subplot(2, 1, 2)
plt.plot(filtered_samples_norm, color='red')
plt.title(f"Filtered Audio Signal (Band-pass: {lowcut}-{highcut} Hz)")

plt.tight_layout()
plt.show()
```
