```python
import numpy as np
import matplotlib.pyplot as plt
def cross_correlation(signal1, signal2):
    # Compute the cross-correlation
    cross_corr = np.correlate(signal1, signal2, mode='full')
    return cross_corr
def autocorrelation(signal):
    # Compute the autocorrelation
    auto_corr = np.correlate(signal, signal, mode='full')
    return auto_corr
# Define the discrete-time signals
signal1 = np.array([1, 2, 3, 4, 5])
signal2 = np.array([2, 4, 6, 8, 10])
# Compute the cross-correlation
cross_corr = cross_correlation(signal1, signal2) 
# Compute the autocorrelation
auto_corr = autocorrelation(signal1) 
# Plot the cross-correlation and autocorrelation signals
plt.figure(figsize=(10, 6))
plt.subplot(2, 1, 1)
plt.stem(cross_corr)
plt.title('Cross-correlation')
plt.xlabel('Time Lag')
plt.ylabel('Magnitude') 
plt.subplot(2, 1, 2)
plt.stem(auto_corr)
plt.title('Autocorrelation')
plt.xlabel('Time Lag')
plt.ylabel('Magnitude')
plt.tight_layout()
plt.show()
```

```python
# Visualize the correlation matrix (sliding window view)
def build_correlation_matrix(signal1, signal2):
    n = len(signal1)
    m = len(signal2)
    total_lags = n + m - 1
    matrix = []
    for lag in range(total_lags):
        # Shift signal2 for current lag
        shifted = np.zeros(n)
        for i in range(m):
            j = lag - (m - 1) + i
            if 0 <= j < n:
                shifted[j] = signal2[i]
        matrix.append(shifted)
    matrix = np.array(matrix)
    # Element-wise multiplication with signal1
    products = matrix * signal1
    # Sum for each lag to get correlation value
    sums = products.sum(axis=1)
    return matrix, products, sums

matrix, products, sums = build_correlation_matrix(signal1, signal2)
print('Correlation Matrix (each row is signal2 shifted):')
print(matrix)
print('Element-wise products:')
print(products)
print('Sum for each lag (correlation result):')
print(sums)
%pip install pandas
# Optionally, display as a table for clarity
import pandas as pd
df = pd.DataFrame(products, columns=[f'x[{i}]' for i in range(len(signal1))])
df['sum'] = sums
display(df)
```

```python
# Visualize the autocorrelation matrix (sliding window view for signal1 with itself)
def build_autocorr_matrix(signal):
    n = len(signal)
    total_lags = 2 * n - 1
    matrix = []
    for lag in range(total_lags):
        shifted = np.zeros(n)
        for i in range(n):
            j = lag - (n - 1) + i
            if 0 <= j < n:
                shifted[j] = signal[i]
        matrix.append(shifted)
    matrix = np.array(matrix)
    products = matrix * signal
    sums = products.sum(axis=1)
    return matrix, products, sums

auto_matrix, auto_products, auto_sums = build_autocorr_matrix(signal1)
print('Autocorrelation Matrix (each row is signal1 shifted):')
print(auto_matrix)
print('Element-wise products:')
print(auto_products)
print('Sum for each lag (autocorrelation result):')
print(auto_sums)

# Optionally, display as a table for clarity
# import pandas as pd
# df_auto = pd.DataFrame(auto_products, columns=[f'x[{i}]' for i in range(len(signal1))])
# df_auto['sum'] = auto_sums
# display(df_auto)
```
